## 2.2 Téléchargement et extraction automatiques des datasets
Cette cellule **télécharge automatiquement** les 17 datasets (UCI + Kaggle), **extrait** les fichiers `.zip` et range tout dans `Dataset ML` (`/content/drive/MyDrive/dataset_ml`) sur Google Drive. Elle est **idempotente** : un dataset déjà présent n'est pas re-téléchargé.

- **UCI** : téléchargement direct des archives `.zip`, extraites dans `_extracted/`.
- **Kaggle** : via `kagglehub` (datasets publics). Si Kaggle demande une authentification, ajouter les secrets Colab `KAGGLE_USERNAME` et `KAGGLE_KEY` (icône 🔑 à gauche).
- Pour D15, seul le fichier `HAM10000_metadata.csv` est téléchargé (pas les ~5 Go d'images).

```python
# =====================================================
# TÉLÉCHARGEMENT + EXTRACTION AUTOMATIQUES -> Dataset ML
# =====================================================
import os, sys, shutil, zipfile, tarfile, subprocess, json, urllib.request, urllib.parse
from pathlib import Path

# 1. Monter Google Drive
from google.colab import drive
drive.mount('/content/drive')

BASE = Path("/content/drive/MyDrive/dataset_ml")      # dossier "Dataset ML"
BASE.mkdir(parents=True, exist_ok=True)
EXTRACT = BASE / "_extracted"                           # zips décompressés ici
EXTRACT.mkdir(exist_ok=True)
ZIPS = BASE / "_zips"                                   # archives brutes conservées
ZIPS.mkdir(exist_ok=True)

# 2. kagglehub + identifiants Kaggle optionnels (secrets Colab)
try:
    import kagglehub
except ImportError:
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "kagglehub"], check=True)
    import kagglehub

try:
    from google.colab import userdata
    for k in ("KAGGLE_USERNAME", "KAGGLE_KEY"):
        try:
            os.environ[k] = userdata.get(k)
        except Exception:
            pass
except Exception:
    pass

# 3. Catalogue : fichier(s) attendu(s) -> source
#    UCI    : (id UCI, slug de l'archive)
#    Kaggle : (propriétaire/dataset, fichier précis ou None pour tout le dataset)
CATALOGUE = {
    "D1":  {"src": "kaggle", "ref": "sid321axn/heart-statlog-cleveland-hungary-final", "file": None,
            "expect": ["heart_statlog_cleveland_hungary_final.csv"]},
    "D2":  {"src": "uci", "id": 45,  "slug": "heart+disease",
            "expect": ["processed.cleveland.data"]},
    "D3":  {"src": "kaggle", "ref": "johnsmith88/heart-disease-dataset", "file": None,
            "expect": ["heart.csv"]},
    "D4":  {"src": "uci", "id": 519, "slug": "heart+failure+clinical+records",
            "expect": ["heart_failure_clinical_records_dataset.csv"]},
    "D5":  {"src": "uci", "id": 60,  "slug": "liver+disorders",
            "expect": ["bupa.data"]},
    "D6":  {"src": "uci", "id": 225, "slug": "ilpd+indian+liver+patient+dataset",
            "expect": ["Indian Liver Patient Dataset (ILPD).csv"]},
    "D7":  {"src": "kaggle", "ref": "fedesoriano/covid19-effect-on-liver-cancer-prediction-dataset", "file": None,
            "expect": ["covid-liver.csv"]},
    "D8":  {"src": "uci", "id": 529, "slug": "early+stage+diabetes+risk+prediction+dataset",
            "expect": ["diabetes_data_upload.csv"]},
    "D9":  {"src": "kaggle", "ref": "abdallamahgoub/diabetes", "file": None,
            "expect": ["diabetes.csv"]},
    "D10": {"src": "kaggle", "ref": "tigganeha4/diabetes-dataset-2019", "file": None,
            "expect": ["diabetes_dataset__2019.csv"]},
    "D11": {"src": "uci", "id": 329, "slug": "diabetic+retinopathy+debrecen",
            "expect": ["messidor_features.arff"]},
    "D12": {"src": "kaggle", "ref": "mansoordaku/ckdisease", "file": None,
            "expect": ["kidney_disease.csv"]},
    "D13": {"src": "kaggle", "ref": "abhia1999/chronic-kidney-disease", "file": None,
            "expect": ["new_model.csv"]},
    "D14": {"src": "kaggle", "ref": "harshghadiya/kidneystone", "file": None,
            "expect": ["kidney-stone-dataset.csv"]},
    "D15": {"src": "kaggle", "ref": "kmader/skin-cancer-mnist-ham10000", "file": "HAM10000_metadata.csv",
            "expect": ["HAM10000_metadata.csv"]},
    "D16": {"src": "uci", "id": 33,  "slug": "dermatology",
            "expect": ["dermatology.data"]},
    "D17": {"src": "uci", "id": 571, "slug": "hcv+data",
            "expect": ["hcvdat0.csv"]},
}

DATA_EXT = {".csv", ".data", ".arff"}

def present(expect):
    """Vrai si tous les fichiers attendus existent déjà quelque part dans BASE."""
    names = {p.name.lower() for p in BASE.rglob("*") if p.is_file()}
    return all(e.lower() in names for e in expect)

def http_get(url, dest, timeout=180):
    req = urllib.request.Request(url, headers={"User-Agent": "Mozilla/5.0"})
    with urllib.request.urlopen(req, timeout=timeout) as r, open(dest, "wb") as f:
        shutil.copyfileobj(r, f)

def extract_archive(archive, target):
    target.mkdir(parents=True, exist_ok=True)
    if zipfile.is_zipfile(archive):
        with zipfile.ZipFile(archive) as zf:
            zf.extractall(target)
    elif tarfile.is_tarfile(archive):
        with tarfile.open(archive) as tf:
            tf.extractall(target)
    else:
        raise ValueError(f"Archive invalide : {archive}")

def uci_candidates(spec):
    """URLs possibles pour l'archive UCI (slug du catalogue, puis slug donné par l'API UCI)."""
    cands = [f"https://archive.ics.uci.edu/static/public/{spec['id']}/{spec['slug']}.zip"]
    try:
        with urllib.request.urlopen(
            urllib.request.Request(f"https://archive.ics.uci.edu/api/dataset?id={spec['id']}",
                                   headers={"User-Agent": "Mozilla/5.0"}), timeout=60) as r:
            name = json.load(r)["data"]["name"]
        slug = urllib.parse.quote_plus(name.lower())
        cands.append(f"https://archive.ics.uci.edu/static/public/{spec['id']}/{slug}.zip")
    except Exception:
        pass
    return cands

def download_uci(did, spec):
    last_err = None
    for url in uci_candidates(spec):
        zpath = ZIPS / f"{did}_{spec['slug'].replace('+','_')}.zip"
        try:
            http_get(url, zpath)
            extract_archive(zpath, EXTRACT / f"{did}_{spec['slug'].replace('+','_')}")
            return True
        except Exception as e:
            last_err = e
            if zpath.exists():
                zpath.unlink()
    raise RuntimeError(f"UCI inaccessible ({last_err})")

def download_kaggle(did, spec):
    dest = BASE / f"{did}_{spec['ref'].split('/')[1]}"
    dest.mkdir(exist_ok=True)
    if spec.get("file"):
        local = kagglehub.dataset_download(spec["ref"], path=spec["file"])
    else:
        local = kagglehub.dataset_download(spec["ref"])
    local = Path(local)
    if local.is_file():
        shutil.copy2(local, dest / local.name)
    else:
        for p in local.rglob("*"):
            if p.is_file():
                rel = p.relative_to(local)
                (dest / rel).parent.mkdir(parents=True, exist_ok=True)
                shutil.copy2(p, dest / rel)
    # extraire d'éventuels zips restants
    for z in list(dest.rglob("*.zip")):
        extract_archive(z, z.with_suffix(""))
    return True

# 4. Boucle de téléchargement (idempotente)
report = {}
for did, spec in CATALOGUE.items():
    if present(spec["expect"]):
        report[did] = "déjà présent"
        print(f"⏭️  {did} : déjà présent sur Drive")
        continue
    try:
        print(f"⬇️  {did} : téléchargement depuis {spec['src'].upper()} ...")
        (download_uci if spec["src"] == "uci" else download_kaggle)(did, spec)
        if present(spec["expect"]):
            report[did] = "téléchargé"
            print(f"✅ {did} : OK")
        else:
            report[did] = "fichier attendu introuvable après extraction"
            print(f"⚠️  {did} : téléchargé mais fichier attendu introuvable : {spec['expect']}")
    except Exception as e:
        report[did] = f"ÉCHEC : {e}"
        print(f"❌ {did} : {e}")

# 5. Bilan
ok = [d for d, s in report.items() if s in ("déjà présent", "téléchargé")]
ko = {d: s for d, s in report.items() if d not in ok}
print(f"\n========== BILAN TÉLÉCHARGEMENT : {len(ok)}/{len(CATALOGUE)} datasets prêts ==========")
for d, s in ko.items():
    print(f"  - {d} : {s}")
if ko:
    print("\n👉 Pour les datasets en échec Kaggle : ajoutez KAGGLE_USERNAME / KAGGLE_KEY dans les secrets Colab "
          "(🔑), puis relancez cette cellule. Les datasets déjà téléchargés ne seront pas re-téléchargés.")

# 6. Lister tous les fichiers de données
all_files = [p for p in BASE.rglob("*")
             if p.is_file() and p.suffix.lower() in DATA_EXT
             and "_zips" not in p.parts and "costs" not in [x.lower() for x in p.parts]]
print(f"\n{len(all_files)} fichiers de données trouvés dans {BASE}")
```

Ces vérifications permettent de détecter les fichiers manquants, les erreurs de chargement et les éventuelles anomalies dans les données.

```python

import warnings
from pathlib import Path

import pandas as pd
from scipy.io import arff

warnings.filterwarnings("ignore")

# =====================================================
# 1. CHEMIN DES DATASETS
# =====================================================

BASE = Path("/content/drive/MyDrive/dataset_ml")

if not BASE.exists():
    raise FileNotFoundError(
        f"Le dossier des datasets est introuvable : {BASE}"
    )

# =====================================================
# 2. NOMBRE DE LIGNES ATTENDU
# =====================================================

EXPECTED_ROWS = {
    "D1": 1190,
    "D2": 303,
    "D3": 1025,
    "D4": 299,
    "D5": 345,
    "D6": 583,
    "D7": 450,
    "D8": 520,
    "D9": 768,
    "D10": 952,
    "D11": 1151,
    "D12": 400,
    "D13": 400,
    "D14": 90,
    "D15": 10015,
    "D16": 366,
    "D17": 615
}

TOTAL_DATASETS = len(EXPECTED_ROWS)

# =====================================================
# 3. CORRESPONDANCE DES FICHIERS
# =====================================================

BY_NAME = {
    "processed.cleveland.data": "D2",
    "heart_failure_clinical_records_dataset.csv": "D4",
    "bupa.data": "D5",
    "indian liver patient dataset (ilpd).csv": "D6",
    "covid-liver.csv": "D7",
    "diabetes_data_upload.csv": "D8",
    "diabetes.csv": "D9",
    "diabetes_dataset__2019.csv": "D10",
    "messidor_features.arff": "D11",
    "kidney_disease.csv": "D12",
    "new_model.csv": "D13",
    "kidney-stone-dataset.csv": "D14",
    "ham10000_metadata.csv": "D15",
    "dermatology.data": "D16",
    "hcvdat0.csv": "D17"
}

# Identification de D1 et D3 par nombre de lignes
ROWS_TO_ID = {
    1190: "D1",
    1025: "D3"
}

# =====================================================
# 4. RECHERCHE DES FICHIERS
# =====================================================

extensions = {".csv", ".data", ".arff"}

all_files = [
    p for p in BASE.rglob("*")
    if p.is_file()
    and p.suffix.lower() in extensions
    and not p.name.lower().startswith("hmnist")
    and p.name.lower() != "o2saturation.csv"
    and "costs" not in [part.lower() for part in p.parts]
]

print(f"Nombre de fichiers trouvés : {len(all_files)}")

# =====================================================
# 5. LECTURE DES FICHIERS CSV
# =====================================================

def read_csv_safe(path, **kwargs):
    try:
        return pd.read_csv(path, **kwargs)
    except UnicodeDecodeError:
        return pd.read_csv(
            path,
            encoding="latin-1",
            **kwargs
        )

# =====================================================
# 6. IDENTIFICATION DES DATASETS
# =====================================================

def identify_dataset(path):

    filename = path.name.lower()

    if filename in BY_NAME:
        return BY_NAME[filename]

    # Identification de HCV même si le fichier
    # possède un suffixe supplémentaire
    if (
        path.suffix.lower() == ".csv"
        and path.stem.lower().startswith("hcvdat0")
    ):
        return "D17"

    # Identification de D1 et D3 par nombre de lignes
    if path.suffix.lower() == ".csv":
        try:
            df = read_csv_safe(path)   # header=0 : len(df) = nombre de patients
            return ROWS_TO_ID.get(len(df))
        except Exception:
            return None

    return None

# =====================================================
# 7. CHARGEMENT DES DATASETS
# =====================================================

def load_dataset(path, dataset_id):

    if dataset_id == "D2":

        columns = [
            "age", "sex", "cp", "trestbps", "chol", "fbs",
            "restecg", "thalach", "exang", "oldpeak",
            "slope", "ca", "thal", "target"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns,
            na_values="?"
        )

    elif dataset_id == "D5":

        columns = [
            "mcv", "alkphos", "sgpt", "sgot",
            "gammagt", "drinks", "selector"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns
        )

    elif dataset_id == "D6":

        columns = [
            "age", "gender", "total_bilirubin",
            "direct_bilirubin", "alkaline_phosphotase",
            "alamine_aminotransferase",
            "aspartate_aminotransferase",
            "total_proteins", "albumin",
            "albumin_globulin_ratio", "dataset"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns,
            na_values="?"
        )

    elif dataset_id == "D16":

        df = read_csv_safe(
            path,
            header=None,
            na_values="?"
        )

        if df.shape[1] == 34:   # 33 attributs + classe

            df.columns = [
                "erythema", "scaling", "definite_borders",
                "itching", "koebner_phenomenon",
                "polygonal_papules", "follicular_papules",
                "oral_mucosal_involvement",
                "knee_elbow_involvement",
                "scalp_involvement", "family_history",
                "melanin_incontinence", "eosinophils_infiltrate",
                "PNL_infiltrate", "fibrosis_papillary_dermis",
                "exocytosis", "acanthosis", "hyperkeratosis",
                "parakeratosis", "clubbing_rete_ridges",
                "elongation_rete_ridges",
                "thinning_suprapapillary_epidermis",
                "spongiform_pustule", "munro_microabcess",
                "focal_hypergranulosis",
                "disappearance_granular_layer",
                "vacuolisation_basal_layer",
                "spongiosis", "saw_tooth_appearance",
                "follicular_horn_plug", "perifollicular_parakeratosis",
                "inflammatory_mononuclear_infiltrate",
                "band_like_infiltrate", "class"
            ]

        return df

    elif dataset_id == "D11":

        raw_data, _ = arff.loadarff(path)
        df = pd.DataFrame(raw_data)

        for col in df.select_dtypes(include=["object"]).columns:
            df[col] = df[col].apply(
                lambda x: x.decode("utf-8")
                if isinstance(x, bytes) else x
            )

        return df

    elif dataset_id == "D17":

        # HCV Hepatitis C Virus
        df = read_csv_safe(path)

        # Suppression des colonnes d'index inutiles
        df = df.loc[
            :,
            ~df.columns.astype(str).str.startswith("Unnamed")
        ]

        return df

    else:

        df = read_csv_safe(path)

        # Suppression des colonnes d'index inutiles
        df = df.loc[
            :,
            ~df.columns.astype(str).str.startswith("Unnamed")
        ]

        return df

# =====================================================
# 8. CHARGEMENT DE TOUS LES DATASETS
# =====================================================

data = {}
loaded_files = {}
errors = []

for file_path in all_files:

    dataset_id = identify_dataset(file_path)

    if dataset_id is None:
        continue

    if dataset_id in data:
        print(f"Attention : {dataset_id} est déjà chargé.")
        continue

    try:

        df = load_dataset(file_path, dataset_id)

        data[dataset_id] = df
        loaded_files[dataset_id] = file_path.name

        print(
            f"OK - {dataset_id} : {file_path.name} "
            f"| Dimensions : {df.shape}"
        )

    except Exception as e:

        errors.append(
            (dataset_id, file_path.name, str(e))
        )

        print(f"ERREUR - {dataset_id} : {e}")

# =====================================================
# 9. BILAN DU CHARGEMENT
# =====================================================

print("\n========== BILAN DU CHARGEMENT ==========")

print(
    f"Datasets chargés : {len(data)}/{TOTAL_DATASETS}"
)

print(
    f"Datasets manquants : {TOTAL_DATASETS - len(data)}"
)

missing = [
    dataset_id
    for dataset_id in EXPECTED_ROWS
    if dataset_id not in data
]

if missing:
    print("Identifiants manquants :", missing)
else:
    print("Tous les datasets D1 à D17 sont chargés.")

# =====================================================
# 10. VERIFICATION DES DIMENSIONS
# =====================================================

print("\n========== VERIFICATION DES DIMENSIONS ==========")

for dataset_id, df in sorted(data.items()):

    expected = EXPECTED_ROWS[dataset_id]
    actual = len(df)

    status = (
        "OK"
        if actual == expected
        else "A VERIFIER"
    )

    print(
        f"{dataset_id} : {actual} lignes "
        f"(attendu : {expected}) --> {status}"
    )

# =====================================================
# 11. DATASETS DISPONIBLES
# =====================================================

print("\nDatasets disponibles :", sorted(data.keys()))
print("L'aperçu de chaque dataset est affiché séparément dans les cellules suivantes (2.4).\n")

# =====================================================
# 12. AFFICHAGE DES ERREURS
# =====================================================

print("\n========== ERREURS DE CHARGEMENT ==========")

if errors:
    for error in errors:
        print(error)
else:
    print("Aucune erreur de chargement.")
```

## 3. Pipeline — Chaîne de traitement des données

Le pipeline de données permet d'organiser et d'automatiser le traitement des 17 datasets médicaux utilisés dans notre projet.

Les données sont récupérées depuis Google Drive, chargées dans Google Colab, puis vérifiées avant leur utilisation dans les étapes suivantes du cycle TDSP.

Le pipeline fonctionne en mode Batch, car les données sont stockées dans des fichiers statiques et traitées lors de l'exécution du notebook.

### 3.1 Choix du type de pipeline

| Critère | Choix | Justification |
|---|---|---|
| Type de traitement | Batch | Les données sont chargées à partir de fichiers existants. |
| Fréquence | À la demande | Le traitement est exécuté lorsque le notebook est lancé. |
| Source | Google Drive | Les fichiers bruts sont stockés dans le dossier `dataset_ml`. |
| Environnement d'exécution | Google Colab | Le notebook Python permet de charger et de traiter les datasets. |
| Organisation | Par identifiant | Chaque dataset est identifié par un ID de D1 à D17. |

### 3.2 Étapes du pipeline

| Étape | Description |
|---|---|
| 1 | Récupération des fichiers depuis Google Drive. |
| 2 | Identification des datasets à partir de leur nom ou de leur structure. |
| 3 | Chargement des fichiers dans des DataFrames pandas. |
| 4 | Vérification des dimensions et du nombre de lignes. |
| 5 | Organisation des datasets dans un dictionnaire. |
| 6 | Transmission des données vers les étapes d'exploration et de préparation. |

### 3.3 Objectifs du pipeline

- Automatiser le chargement des datasets.
- Éviter la répétition du code pour chaque dataset.
- Faciliter la vérification des données.
- Garantir la reproductibilité du traitement.
- Préparer les données pour les étapes suivantes du projet.

**Remarque :** le nettoyage, l'imputation des valeurs manquantes, la normalisation et la binarisation des cibles seront réalisés dans les étapes appropriées du projet.

### 3.4 Automatisation du pipeline

Une fonction unique est utilisée pour parcourir les datasets déjà chargés.

Cette fonction permet de :
- Vérifier la présence des datasets.
- Contrôler leurs dimensions.
- Identifier les valeurs manquantes.
- Compter les doublons.
- Générer un tableau récapitulatif.

Les données originales ne sont pas modifiées durant cette étape.

```python
# =====================================================
# 3.2 PIPELINE AUTOMATISÉ
# =====================================================

import pandas as pd

def execute_pipeline(data):

    results = []

    # Parcours de tous les datasets chargés
    for dataset_id, df in sorted(
        data.items(),
        key=lambda x: int(x[0][1:])
    ):

        results.append({
            "ID": dataset_id,
            "Lignes": df.shape[0],
            "Colonnes": df.shape[1],
            "Valeurs manquantes": int(
                df.isna().sum().sum()
            ),
            "Doublons": int(
                df.duplicated().sum()
            ),
            "Types de variables": str(
                df.dtypes.value_counts().to_dict()
            )
        })

    # Création du tableau récapitulatif
    pipeline_results = pd.DataFrame(results)

    return pipeline_results


# Exécution du pipeline
pipeline_results = execute_pipeline(data)

display(pipeline_results)

print(
    f"Datasets traités : "
    f"{len(pipeline_results)}/17"
)
```
