## 2.2 Chargement des datasets

Nous chargeons les **17 jeux de données (D1 à D17)** utilisés dans notre projet. Les fichiers sont importés depuis Google Drive dans Google Colab.

**Vérification du chargement :** pour chaque dataset, nous vérifions :
- La présence du fichier dans le dossier Google Drive.
- Le bon déroulement du chargement.
- La correspondance du nombre de lignes avec les valeurs attendues dans notre inventaire.
- La présence et l'identification de la colonne cible (*Target*).
- Les dimensions de chaque dataset (nombre de lignes et de colonnes).

Ces vérifications permettent de détecter les fichiers manquants, les erreurs de chargement et les éventuelles anomalies dans les données.

**Remarque :** le dataset D17 (HCV – Hepatitis C Virus) a été ajouté aux 16 datasets initiaux.

```python

import warnings
from pathlib import Path

import pandas as pd
from scipy.io import arff

warnings.filterwarnings("ignore")

# =====================================================
# 1. CHEMIN DES DATASETS
# =====================================================

BASE = Path("/content/drive/MyDrive/dataset_ml")

if not BASE.exists():
    raise FileNotFoundError(
        f"Le dossier des datasets est introuvable : {BASE}"
    )

# =====================================================
# 2. NOMBRE DE LIGNES ATTENDU
# =====================================================

EXPECTED_ROWS = {
    "D1": 1190,
    "D2": 303,
    "D3": 1025,
    "D4": 299,
    "D5": 345,
    "D6": 583,
    "D7": 450,
    "D8": 520,
    "D9": 768,
    "D10": 952,
    "D11": 1151,
    "D12": 400,
    "D13": 400,
    "D14": 90,
    "D15": 10015,
    "D16": 366,
    "D17": 615
}

TOTAL_DATASETS = len(EXPECTED_ROWS)

# =====================================================
# 3. CORRESPONDANCE DES FICHIERS
# =====================================================

BY_NAME = {
    "processed.cleveland.data": "D2",
    "heart_failure_clinical_records_dataset.csv": "D4",
    "bupa.data": "D5",
    "indian liver patient dataset (ilpd).csv": "D6",
    "covid-liver.csv": "D7",
    "diabetes_data_upload.csv": "D8",
    "diabetes.csv": "D9",
    "diabetes_dataset__2019.csv": "D10",
    "messidor_features.arff": "D11",
    "kidney_disease.csv": "D12",
    "new_model.csv": "D13",
    "kidney-stone-dataset.csv": "D14",
    "ham10000_metadata.csv": "D15",
    "dermatology.data": "D16",
    "hcvdat0.csv": "D17"
}

# Identification de D1 et D3 par nombre de lignes
ROWS_TO_ID = {
    1190: "D1",
    1025: "D3"
}

# =====================================================
# 4. RECHERCHE DES FICHIERS
# =====================================================

extensions = {".csv", ".data", ".arff"}

all_files = [
    p for p in BASE.rglob("*")
    if p.is_file()
    and p.suffix.lower() in extensions
    and not p.name.lower().startswith("hmnist")
    and p.name.lower() != "o2saturation.csv"
    and "costs" not in [part.lower() for part in p.parts]
]

print(f"Nombre de fichiers trouvés : {len(all_files)}")

# =====================================================
# 5. LECTURE DES FICHIERS CSV
# =====================================================

def read_csv_safe(path, **kwargs):
    try:
        return pd.read_csv(path, **kwargs)
    except UnicodeDecodeError:
        return pd.read_csv(
            path,
            encoding="latin-1",
            **kwargs
        )

# =====================================================
# 6. IDENTIFICATION DES DATASETS
# =====================================================

def identify_dataset(path):

    filename = path.name.lower()

    if filename in BY_NAME:
        return BY_NAME[filename]

    # Identification de HCV même si le fichier
    # possède un suffixe supplémentaire
    if (
        path.suffix.lower() == ".csv"
        and path.stem.lower().startswith("hcvdat0")
    ):
        return "D17"

    # Identification de D1 et D3 par nombre de lignes
    if path.suffix.lower() == ".csv":
        try:
            df = read_csv_safe(path, header=None)
            return ROWS_TO_ID.get(len(df))
        except Exception:
            return None

    return None

# =====================================================
# 7. CHARGEMENT DES DATASETS
# =====================================================

def load_dataset(path, dataset_id):

    if dataset_id == "D2":

        columns = [
            "age", "sex", "cp", "trestbps", "chol", "fbs",
            "restecg", "thalach", "exang", "oldpeak",
            "slope", "ca", "thal", "target"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns,
            na_values="?"
        )

    elif dataset_id == "D5":

        columns = [
            "mcv", "alkphos", "sgpt", "sgot",
            "gammagt", "drinks", "selector"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns
        )

    elif dataset_id == "D6":

        columns = [
            "age", "gender", "total_bilirubin",
            "direct_bilirubin", "alkaline_phosphotase",
            "alamine_aminotransferase",
            "aspartate_aminotransferase",
            "total_proteins", "albumin",
            "albumin_globulin_ratio", "dataset"
        ]

        return read_csv_safe(
            path,
            header=None,
            names=columns,
            na_values="?"
        )

    elif dataset_id == "D16":

        df = read_csv_safe(
            path,
            header=None,
            na_values="?"
        )

        if df.shape[1] == 35:

            df.columns = [
                "erythema", "scaling", "definite_borders",
                "itching", "koebner_phenomenon",
                "polygonal_papules", "follicular_papules",
                "oral_mucosal_involvement",
                "knee_elbow_involvement",
                "scalp_involvement", "family_history",
                "melanin_incontinence", "eosinophils_infiltrate",
                "PNL_infiltrate", "fibrosis_papillary_dermis",
                "exocytosis", "acanthosis", "hyperkeratosis",
                "parakeratosis", "clubbing_rete_ridges",
                "elongation_rete_ridges",
                "thinning_suprapapillary_epidermis",
                "spongiform_pustule", "munro_microabcess",
                "focal_hypergranulosis",
                "disappearance_granular_layer",
                "vacuolisation_basal_layer",
                "spongiosis", "saw_tooth_appearance",
                "follicular_horn_plug", "perifollicular_parakeratosis",
                "inflammatory_mononuclear_infiltrate",
                "band_like_infiltrate", "class"
            ]

        return df

    elif dataset_id == "D11":

        raw_data, _ = arff.loadarff(path)
        df = pd.DataFrame(raw_data)

        for col in df.select_dtypes(include=["object"]).columns:
            df[col] = df[col].apply(
                lambda x: x.decode("utf-8")
                if isinstance(x, bytes) else x
            )

        return df

    elif dataset_id == "D17":

        # HCV Hepatitis C Virus
        df = read_csv_safe(path)

        # Suppression des colonnes d'index inutiles
        df = df.loc[
            :,
            ~df.columns.astype(str).str.startswith("Unnamed")
        ]

        return df

    else:

        df = read_csv_safe(path)

        # Suppression des colonnes d'index inutiles
        df = df.loc[
            :,
            ~df.columns.astype(str).str.startswith("Unnamed")
        ]

        return df

# =====================================================
# 8. CHARGEMENT DE TOUS LES DATASETS
# =====================================================

data = {}
loaded_files = {}
errors = []

for file_path in all_files:

    dataset_id = identify_dataset(file_path)

    if dataset_id is None:
        continue

    if dataset_id in data:
        print(f"Attention : {dataset_id} est déjà chargé.")
        continue

    try:

        df = load_dataset(file_path, dataset_id)

        data[dataset_id] = df
        loaded_files[dataset_id] = file_path.name

        print(
            f"OK - {dataset_id} : {file_path.name} "
            f"| Dimensions : {df.shape}"
        )

    except Exception as e:

        errors.append(
            (dataset_id, file_path.name, str(e))
        )

        print(f"ERREUR - {dataset_id} : {e}")

# =====================================================
# 9. BILAN DU CHARGEMENT
# =====================================================

print("\n========== BILAN DU CHARGEMENT ==========")

print(
    f"Datasets chargés : {len(data)}/{TOTAL_DATASETS}"
)

print(
    f"Datasets manquants : {TOTAL_DATASETS - len(data)}"
)

missing = [
    dataset_id
    for dataset_id in EXPECTED_ROWS
    if dataset_id not in data
]

if missing:
    print("Identifiants manquants :", missing)
else:
    print("Tous les datasets D1 à D17 sont chargés.")

# =====================================================
# 10. VERIFICATION DES DIMENSIONS
# =====================================================

print("\n========== VERIFICATION DES DIMENSIONS ==========")

for dataset_id, df in sorted(data.items()):

    expected = EXPECTED_ROWS[dataset_id]
    actual = len(df)

    status = (
        "OK"
        if actual == expected
        else "A VERIFIER"
    )

    print(
        f"{dataset_id} : {actual} lignes "
        f"(attendu : {expected}) --> {status}"
    )

# =====================================================
# 11. AFFICHAGE DES DATASETS
# =====================================================

print("\nDatasets disponibles :", sorted(data.keys()))

# Aperçu de chaque dataset
for dataset_id in sorted(data.keys()):

    print(f"\n========== APERCU {dataset_id} ==========")

    display(data[dataset_id].head())

# =====================================================
# 12. AFFICHAGE DES ERREURS
# =====================================================

print("\n========== ERREURS DE CHARGEMENT ==========")

if errors:
    for error in errors:
        print(error)
else:
    print("Aucune erreur de chargement.")
```
